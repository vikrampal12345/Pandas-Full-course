
# Pandas `interpolate()` Notes

### Definition
- `interpolate()` fills missing values (`NaN`) by estimating values from nearby data points.

### Purpose
- Handle missing numerical data without dropping rows.
- Preserve information in the dataset.
- Commonly used in EDA and Time-Series Analysis.

### Syntax

```python
df.interpolate()
```

---

## Important Parameters

### `method='linear'`
- Default method.
- Fills missing values using a straight-line trend.

### `method='nearest'`
- Uses the nearest available value.

### `method='pad'` or `method='ffill'`
- Forward Fill.
- Uses the previous valid value.

### `method='backfill'` or `method='bfill'`
- Backward Fill.
- Uses the next valid value.

### `method='polynomial'`
- Uses a polynomial curve.
- Requires `order`.

### `method='spline'`
- Uses spline interpolation.
- Produces smoother curves.
- Requires `order`.

### `method='time'`
- Interpolates based on time intervals.
- Useful for time-series datasets.

---

## Other Parameters

- `order=n` → Degree of polynomial/spline.
- `axis=0` → Fill column-wise (default).
- `axis=1` → Fill row-wise.
- `limit=n` → Fill only n consecutive NaN values.
- `limit_direction='forward'` → Fill using previous values.
- `limit_direction='backward'` → Fill using next values.
- `limit_direction='both'` → Fill from both directions.
- `inplace=True` → Modify the original DataFrame.

---

## When to Use

✅ Numerical Data  
✅ Time-Series Data  
✅ Sensor Data  
✅ Stock Market Data

## When Not to Use

❌ Categorical Data  
❌ Text Data  
❌ Names, Cities, Labels
