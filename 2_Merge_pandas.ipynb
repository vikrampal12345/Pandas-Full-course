11. Join on Index

Write a Pandas program to merge DataFrames using join() on Index.

```python
import pandas as pd

# Create two sample DataFrames with indexes
df1 = pd.DataFrame({
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Age': [25, 30, 22]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    'Salary': [50000, 60000, 70000]
}, index=[1, 2, 3])


print(df1.join(df2))
```

12. Merge with Indicator

Write a Pandas program to merge with custom indicator to track source.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Age': [25, 30, 22]
})

print(pd.merge(df1, df2, on='ID', how = 'outer', indicator=True))

# indicator show where data comes to merge
```


13. Merge Multiple Keys with Conditions

Write a Pandas program to merge two DataFrames using multiple keys and specific join conditions.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Selena', 'Annabel', 'Charlie', 'Caeso'],
    'City': ['NY', 'LA', 'NY', 'LA']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4, 5],
    'Name': ['Annabel', 'Charlie', 'Caeso', 'Eve'],
    'City': ['LA', 'NY', 'LA', 'NY'],
    'Age': [30, 22, 25, 28]
})

print(pd.merge(df1, df2, how = 'inner', on=['ID', 'City'], indicator=True))
```


15. Merge with Custom Sorting

Write a Pandas program to merge DataFrames with custom sorting.

```python
import pandas as pd
# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 1],
    'Age': [30, 22, 25]
})

merged = pd.merge(df1, df2, on='ID')
print(merged.sort_values(by="Age"))  # using sort_values(by='column_name')
```


16. Merge and Drop Duplicates

Write a Pandas program to merge DataFrames and drop duplicates.

```python
import pandas as pd

# Create two sample DataFrames with potential duplicates
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Annabel', 'Selena', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 1],
    'Name': ['Selena', 'Caeso', 'Annabel'],
    'Age': [30, 22, 25]
})

merged = pd.merge(df1, df2, on=['ID', 'Name'], how='outer')
print(merged)

dub = merged.drop_duplicates()
print(dub)
```


17. Merge with Missing Data

Write a Pandas program to merge DataFrames with missing data.

```python
import pandas as pd

# Create two sample DataFrames with missing data
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [1, 3, 4],
    'Age': [25, 22, 28]
})

merged = pd.merge(df1, df2, on='ID', how='outer')
print(merged)

merged.fillna('unknown', inplace=True)
print(merged)

# merged.fillna(value, inplace=True)
```


18. Merge and Rename Columns

Write a Pandas program to merge DataFrames and rename columns after merge.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [50000, 60000, 70000]
})

merged = pd.merge(df1, df2, on='ID')
print(merged)
merged.rename(columns={'Salary': 'Income'}, inplace = True)
print(merged)
```


19. Merge Multi-Col Diff Names

Write a Pandas program to merge DataFrames on multiple columns with different names.

```python
import pandas as pd

# Create two sample DataFrames with matching values
df1 = pd.DataFrame({
    'Employee_ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']  # Matching names with df2
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Employee_Name': ['Selena', 'Annabel', 'Caeso'],  # Matching names with df1
    'Salary': [50000, 60000, 70000]
})

# Merge the DataFrames on different column names
merged_df = pd.merge(df1, df2, left_on=['Employee_ID', 'Name'], right_on=['ID', 'Employee_Name'])

# Output the result
print(merged_df)
```


20. Merge and Select Columns

Write a Pandas program that merges DataFrames and select specific columns after merge.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['David', 'Annabel', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [50000, 60000, 70000]
})

# Merge the DataFrames on the 'ID' column
merged_df = pd.merge(df1, df2, on='ID')

print(merged_df[['Name', 'Salary']])
```
