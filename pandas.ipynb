# Merge..


```python
# it gives output only the common part
# two or more common part merge into sigle part
import pandas as pd

var1 = pd.DataFrame({"A" : [1,2,3,4,5], "B" : [3,4,5,676,5]})
var2 = pd.DataFrame({"A" : [1,2,3,4,9], "C" : [90,49,39,23,32]})
pd.merge(var1,var2, on="A") # without on="A" it work same as on="A" method
```

```python
import pandas as pd

var1 = pd.DataFrame({"A" : [1,2,3,4,5], "B" : [3,4,5,676,5]})
var2 = pd.DataFrame({"A" : [1,2,3,4,9], "C" : [90,49,39,23,32]})
pd.merge(var1,var2, how="inner") # it is same as the normal
pd.merge(var1, var2, how="left")   # left side data output
pd.merge(var1, var2, how="right")  # right side data output
pd.merge(var1, var2, how="outer" , indicator=True)  # both data 

# indicator=True (give the which data is merge or which side merge)
```

```python
import pandas as pd

var1 = pd.DataFrame({"A" : [1,2,3,4,5], "B" : [3,4,5,676,5]}, index=["a","b","c","d","e"])
var2 = pd.DataFrame({"C" : [21,12,23,49], "D" : [90,49,39,32]},index=["a","b","c","d"])
var2.join(var1)        # it gives ouput according to the var2 DataFrame
var2.join(var1, how="left")    # left side
var2.join(var1, how="right")   # right side
var2.join(var1, how="outer")   # Gives both Union
var2.join(var1, how="inner")   # Gives common data 

```

# Pivot Table and Melt Function = it help reshape the pandas file

# Melt Function

```python
# melt function covert horizontal to vartical like this
import pandas as pd
var1 = pd.DataFrame({"days":[1,1,1,1,2,2],
"eng":[10,12,13,14,15,19],
"math":[23,14,56,56,67,65]
})
var1
pd.melt(var1)
```

```python
pd.melt(var1,id_vars=["eng"]) # convert eng as id
pd.melt(var1,id_vars=["eng"],var_name="python", value_name="number") # change the variable name
```

# Pivot Function = reshaping

```python
import pandas as pd
var1 = pd.DataFrame({"days":[1,1,1,1,5,6],
"st_name":["a","b","c","a","b","c"],
"eng":[10,12,13,14,15,19],
"math":[23,14,56,56,67,65]
})
var1
```

```python
var1.pivot(index="days", columns="st_name")
```

```python
# get particular subject data then 
var1.pivot(index="days", columns="st_name", values="eng")
```

```python
# pivot_table() perform the Arithmatic operation
var1.pivot_table(index="st_name", columns="days", aggfunc="mean") # do sum, median etc
```

```python
# margin by default give the mean
var1.pivot_table(index="st_name", columns="days",margins=True)
```
