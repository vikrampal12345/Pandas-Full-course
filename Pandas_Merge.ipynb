# 2. Outer Join Merge

Write a Pandas program to perform an outer join on two DataFrames.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Age': [25, 30, 22]
})

pd.merge(df1, df2, on='ID', how='outer')   # combine all the columns if missing values then it fill by NaN
```


# 3. Left Join Merge

Write a Pandas program that performs a left join of two DataFrames.

```python
# it join along Left dataframe if available in right with respect left then merge, not then skip the row
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Age': [25, 30, 22]
})

pd.merge(df1, df2, on='ID', how= 'left')
```

# 5. Merge on Multiple Columns

Write a Pandas program to merge two DataFrames on multiple columns.

```python
import pandas as pd

# Create two sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Age': [25, 30, 22]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Salary': [50000, 60000, 70000]
})



pd.merge(df1, df2, on=['ID', 'Name'])  # merge along the both the columns
```

# 6. Merge Overlapping Columns

Write a Pandas program that merges DataFrames with overlapping column names.

```python
import pandas as pd

# Create two sample DataFrames with overlapping column names
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Value': [100, 200, 300]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Value': [400, 500, 600]
})

pd.merge(df1, df2, on=['ID', 'Name'], suffixes=('first', 'seconde'))  # suffix help when the same column name but different value
```


# 7. Merge on Index

Write a Pandas program to merge two DataFrames on their indexes.

```python
import pandas as pd

# Create two sample DataFrames with indexes
df1 = pd.DataFrame({
    'Name': ['Selena', 'Annabel', 'Caeso'],
    'Age': [25, 30, 22]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    'Salary': [50000, 60000, 70000]
}, index=[1, 3, 2])

pd.merge(df1, df2, left_index = True, right_index = True)

# left_index = True means, it allow the left index
# right_index = True mean, it allow the right index
# if both indexes are the same then merge if not then it show the MergeError

```

# 8. Merge Diff Column Names

Write a Pandas program to merge different column names in DataFrames.

```python
import pandas as pd

# Create two sample DataFrames with different join column names
df1 = pd.DataFrame({
    'Employee_ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [1, 3, 2],   # change the sequence but atomatically fix
    'Salary': [50000, 60000, 70000]
})

pd.merge(df1, df2, left_on = 'Employee_ID', right_on = 'ID')
```


# 9. Merge Duplicate Keys

Write a Pandas program to merge DataFrames with duplicate Keys.

```python
import pandas as pd

# Create two sample DataFrames with duplicate keys
df1 = pd.DataFrame({
    'ID': [1, 1, 2],
    'Name': ['Annabel', 'Annabel', 'Selena']
})

df2 = pd.DataFrame({
    'ID': [1, 2],
    'Age': [25, 30]
})

# Merge the DataFrames with duplicate keys
merged_df = pd.merge(df1, df2, on='ID')

# Output the result
print(merged_df)
```

# 10. Merge Multiple DataFrames

Write a Pandas program to merge multiple DataFrames on a common column.

```python

import pandas as pd

# Create three sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Selena', 'Annabel', 'Caeso']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Age': [25, 30, 22]
})

df3 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [50000, 60000, 70000]
})

pd.merge(pd.merge(df1, df2, on="ID"), df3, on="ID")
```
