# Pandas `interpolate()` - Complete Notes & Interview Questions

## Definition
`interpolate()` fills missing values (NaN) by estimating values from nearby data points.

### Purpose
- Handle missing numerical data without dropping rows.
- Preserve information in the dataset.
- Commonly used in EDA and Time-Series Analysis.


````
df.interpolate()
````

## Important Methods

### method='linear'
- Default method.
- Uses a straight-line trend.

### method='nearest'
- Uses the nearest available value.

### method='ffill' or method='pad'
- Uses the previous valid value.

### method='bfill' or method='backfill'
- Uses the next valid value.

### method='polynomial'
- Uses a polynomial curve.
- Requires `order` parameter.

### method='spline'
- Uses a smooth spline curve.
- Requires `order` parameter.

### method='time'
- Interpolates according to time intervals.
- Useful for time-series datasets.


````
# Examples
df.interpolate(method='linear')
df.interpolate(method='nearest')
df.interpolate(method='ffill')
df.interpolate(method='bfill')
df.interpolate(method='polynomial', order=2)
df.interpolate(method='spline', order=3)
df.interpolate(method='time')
````

## Other Parameters

### order=n
- Degree of polynomial/spline interpolation.

### axis=0
- Fill missing values column-wise (default).

### axis=1
- Fill missing values row-wise.

### limit=n
- Fill only n consecutive missing values.

### limit_direction='forward'
- Fill from previous values.

### limit_direction='backward'
- Fill from next values.

### limit_direction='both'
- Fill from both directions.

### inplace=True
- Modify the original DataFrame.


## When to Use

✅ Numerical Data  
✅ Time-Series Data  
✅ Sensor Data  
✅ Stock Market Data  

## When Not to Use

❌ Categorical Data  
❌ Text Data  
❌ Labels / Names


# Interview Questions with Answers

### Q1. What is interpolation?
**Answer:** Interpolation is the process of estimating missing values using surrounding data points.

### Q2. What is the default interpolation method in Pandas?
**Answer:** `method='linear'`.

### Q3. What is the difference between `fillna()` and `interpolate()`?
**Answer:**
- `fillna()` fills NaN with a fixed value.
- `interpolate()` estimates values mathematically.

### Q4. Why use spline interpolation?
**Answer:** Spline interpolation creates smoother curves than linear interpolation.

### Q5. What is the purpose of the `order` parameter?
**Answer:** It specifies the degree of polynomial or spline interpolation.

### Q6. Which interpolation method is useful for time-series data?
**Answer:** `method='time'`.

### Q7. What does nearest interpolation do?
**Answer:** It fills NaN using the closest available value.

### Q8. Can interpolate() be used on categorical data?
**Answer:** No, interpolation is mainly for numerical data.

### Q9. What does the `limit` parameter do?
**Answer:** It restricts the number of consecutive NaN values to fill.

### Q10. What does `inplace=True` do?
**Answer:** It updates the original DataFrame directly.

### Q11. What is the difference between ffill and bfill?
**Answer:**
- `ffill` uses the previous valid value.
- `bfill` uses the next valid value.

### Q12. Which interpolation method is the default and most commonly used?
**Answer:** Linear interpolation.
